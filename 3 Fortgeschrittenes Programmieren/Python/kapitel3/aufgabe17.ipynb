```python
from collections.abc import Generator
from contextlib import closing
from math import sin
import openpyxl

def gen() -> Generator[tuple[float, float]]:
    for x10 in range(-31, 31):
        x = x10 / 10
        y = x * sin(x)
        yield x, y

def main() -> None:
    # Open the workbook
    # The closing() automatically calls the close() method of the Workbook, since Workbook does not support context managers by itself
    with closing(openpyxl.Workbook()) as workbook:
        # Change the title active sheet
        worksheet = workbook.active
        worksheet.title = 'MeineErsteTabelle'
        
        # Write headings for table
        worksheet['A1'] = 'x'
        worksheet['B1'] = 'y'
        
        for i, (x, y) in enumerate(gen(), 2):
            worksheet['A' + str(i)] = x
            worksheet['B' + str(i)] = y
    
        # Save to aufgabe17.xlsx
        workbook.save('aufgabe17.xlsx')

main()
```
