```python
import re

def filterHTML(src: str, dest: str) -> None:  # noqa Invalid name
    with open(src, 'r') as src_file:
        html = src_file.read()

    # < matches character <
    # [^>] matches any character other than > one or more times
    # > matches the character >
    # .+ cannot be used because it would match everything until the last occurrence of >
    # re.sub replaces any occurrence of the pattern
    text = re.sub(r'<[^>]+>', '', html)

    with open(dest, 'w') as dest_file:
        dest_file.write(text)
```

```python
src = "aufgabe15.html"
dest = "aufgabe15.txt"

filterHTML(src, dest)
```
