## Übungsaufgabe: Nutzerverwaltungssystem

Erstellen Sie ein **einfaches Nutzerverwaltungssystem** in Python, das folgende Anforderungen erfüllt:

1. **Nutzerprofil**: Definieren Sie eine Funktion ```erstelle_nutzerprofil```, die ein Nutzerprofil als Dictionary erstellt. Das Profil soll folgende Informationen enthalten:
- Vorname (String)
- Nachname (String)
- Alter (Integer)

2. **Nutzer hinzufügen**: Erstellen Sie eine Liste ```nutzer_liste```, in die mehrere Nutzerprofile hinzugefügt werden können. Verwenden Sie dabei eine Schleife, die dreimal aufgerufen wird, sodass zwei Nutzerprofile nacheinander eingegeben und gespeichert werden.

3. **Filterfunktion**: Erstellen Sie eine Funktion ```filtern_nutzer```, die mithilfe einer Lambda-Funktion und der filter()-Funktion alle Nutzer ab einem bestimmten Alter (z.B. 18 Jahre) zurückgibt.

4. **Ausgabe**: Geben Sie die gefilterten Nutzer und das Set der Interessen als formatierten String aus.

5. **Speichern**: Speichern Sie die gefilterten Nutzer in einer Textdatei ```nutzer_ausgabe.txt``` und lesen Sie deren Inhalt anschließend aus, um zu überprüfen, ob die Speicherung erfolgreich war.

**Hinweise**

- Verwenden Sie ```input()``` für die Eingaben (z.B. Vorname, Nachname, Alter, Interessen).
- Formatieren Sie die Ausgabe so, dass die Nutzerdaten gut lesbar sind.

````python
import json
from typing import Any


def erstelle_nutzerprofil() -> dict[str, Any]:
    vorname = input('Vorname: ')
    nachname = input('Nachname: ')
    alter = int(input('Alter: '))
    
    return {
        'vorname': vorname,
        'nachname': nachname,
        'alter': alter
    }

nutzer_liste: list[dict[str, Any]] = [erstelle_nutzerprofil() for _ in range(3)]

def filter_nutzer(min_alter: int) -> list[dict[str, Any]]:
    return list(filter(lambda n: n['alter'] >= min_alter, nutzer_liste))

for nutzer in filter_nutzer(18):
    print(f'{nutzer['vorname']} {nutzer['nachname']}, {nutzer["alter"]} Jahre alt')

with open('nutzer_liste.json', 'w') as outfile:
    json.dump(filter_nutzer(18), outfile)
````
