Gegeben sind die beiden Listen `keys` und `values`. Erzeugen Sie daraus ein Dictionary mittels einer Schleife, mit deren Hilfe Sie durch die Listen laufen. Testen Sie Ihren Code mit:

```python
keys = ['a', 'b', 'c', 'd']
values = [1, 2, 3, 4]

dict_loop = {}

for key, value in zip(keys, values):
    dict_loop[key] = value
    
print(dict_loop)
```

Machen Sie nochmal dasselbe, aber mit "Dict Comprehension".

```python
dict_compr = {key: value for key, value in zip(keys, values)}
print(dict_compr)
```

Gegeben sind zwei Listen `a` und `b`. Tauschen Sie darin alle Elemente aus, sodass schließlich `a` alle Elemente von `b` enthält und umgekehrt. Testen Sie Ihren Code mit

```python
from copy import deepcopy

a = [1, 2, 3, 4, 5]
b = [6, 7, 8, 9, 10]

temp = deepcopy(a)
a = deepcopy(b)
b = temp
del temp

print(f'{a=}, {b=}')
```

Gegeben ist eine Liste von Zahlen. Erzeugen Sie eine neue Liste, die nur Werte größer 3 enthält. Lösen Sie dies zunächst mit einer Schleife. Verwenden Sie dann List Comprehension. 

```python
x = [4, 3, 2, 8, 6, 0, 4, 6, -2, 1]

filtered_loop = []

for item in x:
    if item > 3:
        filtered_loop.append(item)
print(filtered_loop)
```

```python
filtered_compr = [i for i in x if i > 3]
print(filtered_compr)
```

Erstellen Sie eine tiefe Kopie einer Liste von Listen von Zahlen. Dabei darf das Ändern einer Zahl in der Kopie nicht das Original verändern. Testen Sie Ihren Code mit

```python
l = [[1, 2, 3], [7, 6, 5, 4], [8, 9]]

l2 = deepcopy(l)
l2[0][0] = 1279

print(f'{l=}')
print(f'{l2=}')
```
