# Aufgabe 5

Sie finden an Ihrem Kühlschrank Magneten mit unten stehenden Code-Stückchen vor. Können Sie die Schnipsel wieder so anordnen, dass sie ein funktionierendes Python-Programm ergeben, das folgende Ausgabe ergibt:

```python
Insel: Fidschi
Insel: Kuba
Insel: Bermudas
Insel: Komoren
```

Fügen Sie so viele Einrückungen dazu, wie Sie benötigen. Erläutern Sie kurz die Funktionsweise des Programms.

<img src="aufgabe5.png" alt="Alternative text" />

````python
inseln = ['Bermudas', 'Fidschi', 'Komoren', 'Kuba']
index = [1, 3, 0, 2]

y = 0
ref = 0

while y < 4:
    ref = index[y]
    print(f'Insel: {inseln[ref]}')
    y += 1
````
