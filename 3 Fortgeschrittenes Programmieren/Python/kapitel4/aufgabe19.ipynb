```python
class SparseVector:
    def __init__(self) :
        self.__vec = {}
    def __getitem__(self, index):   
        return self.__vec.get(index, 0.0)

    def __setitem__(self, index, value):
        self.__vec[index] = value

s = SparseVector()

s[11] = 3.14
s[17] = 6.28

for i in range(10, 20) :
    print("Wert[", i, "] = ", s[i], end = ", ")
```
