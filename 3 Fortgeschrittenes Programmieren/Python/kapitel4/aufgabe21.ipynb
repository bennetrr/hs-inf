```python
class Bauland:
    _preise: dict[str, tuple[str, int]] = {
        'BA': ('Bamberg', 218.95),
        'BT': ('Bayreuth', 180.12),
        'HO': ('Hof', 72.86),
        'CO': ('Coburg', 70.84)
    }
    
    __flaeche: float
    _kennz: str
    
    def __init__(self, kennz: str, flaeche: float) -> None:
        self.__flaeche = float(flaeche)
        self._kennz = kennz
        
    def getFlaeche(self) -> float:
        return self.__flaeche
    
    def getPreis(self) -> float:
        return self._preise[self._kennz][1] * self.__flaeche
    
    def __str__(self) -> str:
        return f'Ein Baugrundstück in {self._preise[self._kennz][0]} mit einer Fläche von {self.__flaeche:.2f} qm'
```

```python
grundst = Bauland('CO', 1200)
print('Fläche: ', grundst.getFlaeche(), ' qm')
print('Preis: ', grundst.getPreis(), ' EUR')
```

```python
class Grundstueck(Bauland):
    __laenge: float
    __breite: float
    
    def __init__(self, kennz: str, laenge: float, breite: float) -> None:
        self.__laenge = float(laenge)
        self.__breite = float(breite)
        
        super().__init__(kennz, laenge * breite)
        
    def __str__(self) -> str:
        return f'Ein Baugrundstück in {self._preise[self._kennz][0]}, {self.__laenge:.2f} m mal {self.__breite:.2f} m'
```
