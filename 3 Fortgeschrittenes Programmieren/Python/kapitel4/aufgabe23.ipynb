```python
class Kunde:
    _tracker: int = 0
    
    def __init__(self, nr = 0, name =''):
        Kunde._tracker += 1
        self.__nummer = nr
        self.__name = name
        
    def __del__(self):
        Kunde._tracker -= 1
        
    @classmethod
    def getAnzahl(cls) -> int:
        return cls._tracker

    def getNummer(self):
        return self.__nummer
    
    def getName(self) :
        return self.__name
    
    def __str__(self) :
        return f'{str(self.__nummer)}: {self.__name}'
```

```python
# Erstelle einen Kunden ohne Angabe von Details
kunde1 = Kunde()

# Test 1: Überprüfe die Nummer des Kunden (sollte 0 sein)
assert kunde1.getNummer() == 0, "Test 1 fehlgeschlagen!"

# Test 2: Überprüfe den Namen des Kunden (sollte leer sein)
assert kunde1.getName() == "", "Test 2 fehlgeschlagen!"

# Test 3: Überprüfe die String-Repräsentation des Kunden
assert str(kunde1) == "0: ", "Test 3 fehlgeschlagen!"

# Test 4: Überprüfe die Gesamtanzahl der Kunden (sollte 1 sein)
assert Kunde.getAnzahl() == 1, "Test 4 fehlgeschlagen!"

# Erstelle einen weiteren Kunden mit Details
kunde2 = Kunde(1001, "Max Mustermann")

# Test 5: Überprüfe die Nummer des Kunden
assert kunde2.getNummer() == 1001, "Test 5 fehlgeschlagen!"

# Test 6: Überprüfe den Namen des Kunden
assert kunde2.getName() == "Max Mustermann", "Test 6 fehlgeschlagen!"

# Test 7: Überprüfe die String-Repräsentation des Kunden
assert str(kunde2) == "1001: Max Mustermann", "Test 7 fehlgeschlagen!"

# Test 8: Überprüfe die Gesamtanzahl der Kunden (sollte 2 sein)
assert Kunde.getAnzahl() == 2, "Test 8 fehlgeschlagen!"

# Lösche einen Kunden
del kunde1

# Test 9: Überprüfe die Gesamtanzahl der Kunden nach dem Löschen (sollte 1 sein)
assert Kunde.getAnzahl() == 1, "Test 9 fehlgeschlagen!"

print("Alle Tests erfolgreich durchgelaufen!")
```
