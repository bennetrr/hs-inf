```python
from functools import reduce

def tag_des_jahres(tag: int, monat: int, jahr: int):
    schaltjahr = ((jahr % 4 == 0) and (jahr % 100 != 0)) or (jahr % 400 == 0)
    
    tage_pro_monat = [
        0,
        31,  # Januar
        29 if schaltjahr else 28,  # Februar
        31,  # März
        30,  # April
        31,  # Mai
        30,  # Juni
        31,  # Juli
        31,  # August
        30,  # September
        31,  # Oktober
        30,  # November
        31  # Dezember
    ]
    
    if monat > 12 or tag > tage_pro_monat[monat] or jahr < 2000:
        print('Ungültiges Datum')
        return 0
    
    tage_in_monaten_davor = reduce(lambda x, y: x + y, tage_pro_monat[:monat], 0)
    tdj = tage_in_monaten_davor + tag

    print(f'Der {tag:02d}.{monat:02d}.{jahr} ist der {tdj}. Tag des Jahres')
    return tdj
```

```python
from datetime import datetime

def tag_des_jahres(tag: int, monat: int, jahr: int):
    try:
        dt = datetime(jahr, monat, tag)
    except ValueError:
        print('Ungültiges Datum')
        return 0

    tdj = dt.timetuple().tm_yday
    
    print(f'Der {tag:02d}.{monat:02d}.{jahr} ist der {tdj}. Tag des Jahres')
    return tdj
```

```python
assert tag_des_jahres(15, 3, 2023) == 74, "Testfall 1 fehlgeschlagen"
assert tag_des_jahres(31, 12, 2023) == 365, "Testfall 2 fehlgeschlagen"
assert tag_des_jahres(29, 2, 2024) == 60, "Testfall 3 fehlgeschlagen"
assert tag_des_jahres(32, 1, 2023) == 0, "Testfall 4 fehlgeschlagen"
assert tag_des_jahres(1, 13, 2023) == 0, "Testfall 5 fehlgeschlagen"
assert tag_des_jahres(1, 1, 1023) == 0, "Testfall 6 fehlgeschlagen"
```
