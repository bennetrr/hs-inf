```python
class Ding:
    _dichte: dict[str, tuple[str, float]] = {
        'Au': ('Gold', 19.32),
        'Ag': ('Silber', 10.5),
        'Fe': ('Eisen', 7.87)
    }

    __volumen: float
    _symbol: str

    def __init__(self, symbol: str, volumen: int) -> None:
        if symbol not in Ding._dichte:
            print(f'Symbol {symbol} nicht bekannt.')
            raise Exception
        self._symbol = symbol
        self.__volumen = volumen

    def getMasse(self) -> float:
        return self._dichte[self._symbol][1] * self.__volumen

    def getVolumen(self) -> float:
        return self.__volumen
    
    def __str__(self) -> str:
        return f'Ein Ding aus {self._dichte[self._symbol][0]} mit einer Masse von {self.getMasse():.2f} g'
```

```python
# Instanz mit Gold erstellen
gold_ding = Ding('Au', 100)

# Test 1: Überprüfe das Volumen
assert gold_ding.getVolumen() == 100, "Test 1 fehlgeschlagen!"

# Test 2: Überprüfe die Masse
# Erwartete Masse: 100 * 19.32
assert gold_ding.getMasse() == 1932.0, "Test 2 fehlgeschlagen!"

# Test 3: Überprüfe die String-Repräsentation
print(gold_ding)
```
