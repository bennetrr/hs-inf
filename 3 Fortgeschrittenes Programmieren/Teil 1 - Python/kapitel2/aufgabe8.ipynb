```python
def rechne(func, *nums):
    return tuple(func(num) for num in nums)
```

```python
def sq(z):
    return z*z

assert rechne(sq, 1, 3, 4, 7) == (1, 9, 16, 49)
```

```python
def verdoppeln(x):
    return x * 2

assert rechne(verdoppeln) == ()
assert rechne(len, 'apple', 'banana', 'cherry') == (5, 6, 6)
```

```python
def konvertiere_zu_string(x):
    return str(x)

assert rechne(konvertiere_zu_string, 1, 2.3, 'abc', True) == ('1', '2.3', 'abc', 'True')
```
