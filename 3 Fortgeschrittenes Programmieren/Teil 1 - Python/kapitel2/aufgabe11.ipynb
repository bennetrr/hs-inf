```python
heim = {'Andi', 'Ben', 'Carl', 'Daniel'}
gast = {'Emil', 'Finn', 'Gerd', 'Holger'}
star = {'Rafael'}

paare = {
    *[frozenset((g, h)) for h in heim for g in gast],
    *[frozenset((h, list(star)[0])) for h in heim],
    *[frozenset((g, list(star)[0])) for g in gast]
}

paare = {tuple(paar) for paar in paare}
```

```python
for i, p in enumerate(sorted(paare)):
    print(f'{i + 1}: {p}')
```
