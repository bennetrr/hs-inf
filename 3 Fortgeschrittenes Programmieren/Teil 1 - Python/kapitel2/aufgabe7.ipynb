```python
def wochentag(tag, monat, jahr):
    schaltjahr = ((jahr % 4 == 0) and (jahr % 100 != 0)) or (jahr % 400 == 0)
    
    tage_pro_monat = [
        0,
        31,  # Januar
        29 if schaltjahr else 28,  # Februar
        31,  # März
        30,  # April
        31,  # Mai
        30,  # Juni
        31,  # Juli
        31,  # August
        30,  # September
        31,  # Oktober
        30,  # November
        31  # Dezember
    ]
    
    if monat > 12 or tag > tage_pro_monat[monat]:
        print(f'{tag:02}.{monat:02}.{jahr} ist ein ungültiges Datum')
        return 'ungültig'
    
    if monat <= 2:
        monat += 12
        jahr -= 1
    
    c = jahr // 100  # Jahrhundert
    y = jahr % 100   # Jahr im Jahrhundert
    
    # Zeller-Formel für den Wochentag
    wtn = (tag + (26 * (monat + 1) // 10) + y + (y // 4) + (c // 4) + (5 * c)) % 7
    
    # Die Zeller-Formel gibt 0 für Samstag, 1 für Sonntag, etc., daher muss die Zuordnung angepasst werden.
    wt = {
        0: 'Samstag',
        1: 'Sonntag',
        2: 'Montag',
        3: 'Dienstag',
        4: 'Mittwoch',
        5: 'Donnerstag',
        6: 'Freitag'
    }[wtn]

    print(f"Der {tag:02}.{monat - 12 if monat > 12 else monat:02}.{jahr + 1 if monat > 12 else jahr} ist ein {wt}.")
    return wt
```

```python
from datetime import datetime

def wochentag(tag, monat, jahr):
    try:
        dt = datetime(jahr, monat, tag)
    except ValueError:
        print('Ungültiges Datum')
        return 'ungültig'
    
    wtn = dt.isoweekday()
    
    wt = {
        1: 'Montag',
        2: 'Dienstag',
        3: 'Mittwoch',
        4: 'Donnerstag',
        5: 'Freitag',
        6: 'Samstag',
        7: 'Sonntag',
    }[wtn]

    print(f"Der {tag:02}.{monat - 12 if monat > 12 else monat:02}.{jahr + 1 if monat > 12 else jahr} ist ein {wt}.")
    return wt
```

```python
assert wochentag(29, 2, 2000) == "Dienstag"
assert wochentag(1, 3, 2000) == "Mittwoch"
assert wochentag(1, 1, 2000) == "Samstag"
assert wochentag(23, 10, 23) == "Montag"
assert wochentag(29, 2, 2001) == "ungültig"
assert wochentag(32, 1, 2000) == "ungültig"
assert wochentag(1, 13, 2022) == "ungültig"
```
