# `deepcopy` vs `copy`

`copy` fertigt eine "shallow copy" (= oberflächlich / flach) an.
Das heißt, dass eine neue Liste mit allen Elementen der Ursprungsliste erstellt wird.

Im Gegensatz dazu arbeitet `deepcopy` rekursiv: Bei Listen in Listen wird die Unterliste auch kopiert und nicht nur referenziert.

Das kann man einfach anhand der Identity eines Objektes (in CPython die Speicheradresse) zeigen:

```python
from copy import copy, deepcopy

list_original = [[1, 2, 3], [7, 6, 5, 4], [8, 9]]

list_referenced = list_original
list_shallow = copy(list_original)
list_deep = deepcopy(list_original)

print(f'Identity of list_original:       {id(list_original):#x}')
print(f'Identity of list_referenced:     {id(list_referenced):#x}, same as list_original: {list_original is list_referenced}')
print(f'Identity of list_shallow:        {id(list_shallow):#x}, same as list_original: {list_original is list_shallow}')
print(f'Identity of list_deep:           {id(list_deep):#x}, same as list_original: {list_original is list_deep}')
```

```python
print(f'Identity of list_original[0]:    {id(list_original[0]):#x}')
print(f'Identity of list_referenced[0]:  {id(list_referenced[0]):#x}, [0] same as list_original[0]: {list_original[0] is list_referenced[0]}')
print(f'Identity of list_shallow[0]:     {id(list_shallow[0]):#x}, [0] same as list_original[0]: {list_original[0] is list_shallow[0]}')
print(f'Identity of list_deep[0]:        {id(list_deep[0]):#x}, [0] same as list_original[0]: {list_original[0] is list_deep[0]}')
```

## Vereinfachte Implementation für `copy`

```python
def copy(l):
    return [i for i in l]
```

## Vereinfachte Implementation für `deepcopy`

```python
def deepcopy(l):
    return [
        deepcopy(i) if isinstance(i, list) else i for i in l
    ]
```
