# Aufgabe 4

Wir wollen den Median einer Liste von Zahlen bestimmen. Dieser ist der mittlere Wert einer Liste, wobei die Zahlen aufsteigend sortiert sein müssen. Der Median von fünf Ganzzahlen ist z.B. bei 7, 3, 2, 5, 9 die 5.  

Lassen Sie also zunächst vom Benutzer die Anzahl der Zahlen und die Werte selbst eingeben.

```python
anzahl_zahlen = int(input('Anzahl Zahlen: '))

zahlen = [int(input(f'Zahl {i+1}')) for i in range(anzahl_zahlen)]
print(zahlen)
```

Sortieren Sie nun die Liste und geben Sie den Median aus.

```python
zahlen_sortiert = sorted(zahlen)
mitte = len(zahlen) // 2
median = zahlen_sortiert[mitte]

print(f'{median=}')
```
