## 1.2 Ganzzahldivision

Lassen Sie sich zwei Zahlen vom Benutzer eingeben. Geben Sie dann die Division 'erste Zahl geteilt durch zweite' aus. Verwenden Sie auch die Ganzzahldivision und geben Sie den Rest der Division aus. Vermeiden Sie einen `ZeroDivisionError`.

```python
x = int(input('x = '))
y = int(input('y = '))

print(f'{x / y = }' if y != 0 else 'x/y = ∞')
print(f'{x // y = }')
print(f'{x % y = }')
```

## 1.3 Große Zahlen

Ihre neue Festplatte hat 1 TiB. Wie viele Byte sind das genau? Davon werden 1,5 % für das Dateisystem benötigt. Wie viele Byte stehen also zur Verfügung?

```python
size_in_tib = 1
size_in_byte = size_in_tib * (2**10)**4
available_space = size_in_byte - size_in_byte * 0.015

print(f'{size_in_tib} TiB = {size_in_byte} B')
print(f'Available Space = {available_space} B')
```
