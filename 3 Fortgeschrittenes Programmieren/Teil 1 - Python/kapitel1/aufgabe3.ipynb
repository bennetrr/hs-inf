# Aufgabe 3

Wir wollen die Verzinsung eines Kapitals nach einigen Jahren betrachten. Lassen Sie sich zunächst den Betrag des Kapitals, den Zinssatz in Prozent sowie die Laufzeit in Jahren eingeben und wandeln Sie Kapital und Zinssatz nach `float`, die Laufzeit nach `int` um:

```python
kapital = float(input('Kapital (€): '))
zinssatz = float(input('Zinssatz (%): '))
laufzeit = int(input('Laufzeit (Jahre): '))
```

Berechnen Sie nun den Endbetrag und geben Sie alle Informationen aus.

```python
endbetrag = kapital * (1 + (zinssatz / 100)) ** laufzeit
print(f'Endbetrag = {endbetrag}')
```
