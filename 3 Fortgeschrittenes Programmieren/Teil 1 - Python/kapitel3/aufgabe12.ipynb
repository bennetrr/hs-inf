```python
from math import pi

def gm(l: list[float]) -> list[float]:
    return list(map(lambda n: n * 360 / (2*pi), filter(lambda n: 0 <= n <= 2*pi, l)))
```

```python
q = [-1.57, 0.1, 1.4, 2.3, 2.6, 2.9, 3.14, 4.0, 17.0]
assert str(gm(q)) == '[5.729577951308232, 80.21409131831524, 131.78029288008932, 148.96902673401405, 166.15776058793872, 179.90874767107852, 229.1831180523293]'
```
