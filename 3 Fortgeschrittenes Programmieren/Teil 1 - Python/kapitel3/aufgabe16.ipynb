```python
from collections.abc import Generator
from pathlib import Path
from datetime import datetime, timedelta

def main() -> None:
    root_dir = Path(input('Wurzel des Verzeichnisbaums: '))
    changed_since = timedelta(hours=int(input('Zeitintervall (Stunden): ')))
    
    files = list_files(root_dir, changed_since)
    
    print('Datei und Zeitpunkt der letzten Änderung:')
    
    for file in files:
        path: Path = file[0]
        changed: datetime = file[1]
        
        print(path, changed.isoformat())

def list_files(root_dir: Path, changed_since: timedelta) -> list[tuple[Path, datetime]]:
    # Get all files recursively
    files: Generator[Path] = root_dir.rglob('*')
    min_datetime = datetime.now() - changed_since
    
    res = []
    
    for file in files:
        # Read the last modified time attribute of the file and parse it as a timestamp
        changed = datetime.fromtimestamp(file.stat().st_mtime)
        
        if changed >= min_datetime:
            res.append((file, changed))
            
    return res

main()
# list_files(Path('/Users/wemogymac/projects-hs/'), timedelta(hours=2))
```
